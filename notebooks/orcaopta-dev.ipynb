# 7. Visualization libraries
# -----------------------------

```python

import networkx as nx
from pyvis.network import Network

G = nx.DiGraph()
G.add_node("OpenStack")
G.add_node("Kubernetes")
G.add_edge("OpenStack", "Kubernetes")

net = Network(notebook=True)
net.from_nx(G)
net.show("cloud_graph.html")

print("Interactive cloud graph generated: cloud_graph.html")
```
